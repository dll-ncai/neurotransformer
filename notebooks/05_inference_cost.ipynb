# 5. Model size and inference cost

Parameters, FLOPs per 2 s window, and single-window latency on CPU and (if available) GPU for Neurotransformer
and its ablation variants. FLOPs are counted with `thop` (FLOPs = 2 × MACs). Latency depends on your hardware.

```python
import pandas as pd

from neurotransformer.benchmark import model_costs
from neurotransformer.models import MODEL_REGISTRY, build_model

N_CLASSES = 3
rows = [{"model": name, **model_costs(build_model(name, N_CLASSES), window_samples=400, iters=100)}
        for name in MODEL_REGISTRY]
costs = pd.DataFrame(rows).set_index("model")
costs.round(3)
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col, title in zip(axes, ["params_m", "mflops_per_segment", "cpu_latency_ms"],
                          ["Parameters (M)", "MFLOPs per window", "CPU latency (ms)"]):
    costs[col].plot.bar(ax=ax, rot=45, title=title)
plt.tight_layout()
```
